# **The Price of Ore: Do Strong Institutions Protect Environmental Policy from Mineral Dependence?**

## 1. Introduction

#### Do mineral-dependent countries have lower environmental tax revenues, and does government effectiveness condition this relationship?

Do mineral-dependent countries have weaker environmental regulations, and does government effectiveness explain why some escape this fate? This study investigates whether institutional quality conditions the relationship between mineral wealth and environmental policy. We use environmental tax revenue as a percentage of GDP (IMF) as a proxy for environmental refulations and the key independent variable, mineral rents as a percentage of GDP (World Bank), captures a country's economic dependence on the mining sector. The core hypothesis is that mineral dependence is associated with lower environmental taxes, but this relationship is not deterministic: strong government institutions may shield environmental policy from mining sector pressure.

The mechanism Z is government effectiveness which is drawn from the World Bank's Worldwide Governance Indicators. Government effectiveness measures the quality of public services, the competence of the civil service, and the credibility of government commitments to policies. In weak institutional environments, mining companies can evade taxes, capture regulators, and lobby successfully against environmental protections. In strong institutional environments, professional civil servants enforce rules impartially, policy changes require legislative approval, and contracts cannot be secretly renegotiated. Thus, government effectiveness is hypothesized to moderate the mineral dependence-environmental tax relationship: high-effectiveness countries should exhibit a weaker (or absent) negative association, while low-effectiveness countries should show the strongest deterioration in environmental policy.

This relationship is tested while controlling for a comprehensive set of explanatory variables that theory suggests influence environmental policy. Richer economies (GDP per capita) may demand stricter environmental regulation through an Environmental Kuznets Curve mechanism. Larger government consumption proxies state capacity to design and enforce tax policy. Industrial value added controls for the political influence of manufacturing and extraction lobbies. Trade openness captures globalization pressures that may trigger races-to-the-bottom. Urban population share reflects concentrated citizen demand for environmental quality. Finally, additional governance indicators—political stability and voice and accountability—account for other dimensions of institutional quality. The control variables ensure the core relationship between mineral dependence, government effectiveness, and environmental taxes is not confounded by broader economic, structural, or institutional factors.


### 1.1.1 Methodology 

The analysis relies on merging two datasets covering the period 2003–2020. The dependent variable, environmental tax revenue as a percentage of GDP, is sourced from the International Monetary Fund's Climate Change Indicators Dashboard. This measure captures the government's fiscal commitment to taxing pollution and resource use, serving as a proxy for environmental regulatory stringency.

The key explanatory variable, mineral rents as a percentage of GDP, is drawn from the World Bank's World Development Indicators (WDI), along with the mechanism variable government effectiveness, accompanied with comprehensive set of economic and institutional controls. Each control variable is included based on established theoretical channels linking it to environmental policy outcomes. GDP per capita accounts for the Environmental Kuznets Curve hypothesis that wealthier economies can afford and demand stricter environmental protection. GDP growth captures business cycle effects that may temporarily relax or tighten fiscal policy priorities. General government consumption as a share of GDP proxies state capacity—governments with greater administrative resources are better equipped to design, implement, and enforce environmental tax systems. Industry value added controls for the political influence of manufacturing and extraction sectors, which typically mobilize against policies that increase production costs. Trade openness captures two opposing globalization pressures: races-to-the-bottom where governments weaken environmental policy to attract mobile capital, or alternatively, the diffusion of stricter environmental norms through trade partnerships. Urban population share reflects concentrated citizen demand for environmental quality, as urban voters experience pollution more directly and can more effectively organize around shared concerns. Finally, the World Governance Indicators—government effectiveness, political stability, and voice and accountability—measure institutional quality and democratic constraints that determine whether policy can be captured by resource interests or remains responsive to broader public goods.


### 1.1.2 Literature review 

The relationship between natural resource wealth and environmental outcomes has received considerable attention in the economics literature, yet the policy mechanisms linking these variables remain poorly understood. A substantial body of research establishes that commodity price booms exacerbate environmental degradation. Oyekola, Emediegwu, and Animashaun (2024) find that positive commodity price shocks significantly increase CO₂ emissions across 179 countries, with effects concentrated in democratic regimes where interest group competition intensifies pressure on environmental outcomes. Similarly, Chen et al. (2024) demonstrate that natural resource rents contribute to higher emissions in emerging economies, while Danish, Ulucak, and Baloch (2023) confirm that resource dependence increases pollution in the United States through both direct extraction channels and indirect effects on energy intensity. These studies collectively suggest that resource abundance, particularly during price upswings, carries environmental costs.

The mechanisms through which resource dependence affects environmental quality, however, remain contested. One strand of literature emphasizes technological and regulatory factors as potential mitigators. Luo and Mabrouk (2022) show that green technology innovation and environmental policy stringency can offset resource-driven ecological damage in resource-rich countries, reducing ecological footprints even as extraction expands. Dai et al. (2024) extend this finding to major emitters, demonstrating that while mineral resources generally impede sustainable development, the magnitude of this effect varies considerably across countries and depends on institutional context. Shahbaz et al. (2019) similarly find that globalization and financial development moderate the environmental impact of natural resources in top-emitting nations.

Another strand focuses on institutional and political economy explanations. Espin and Perz (2021) provide qualitative evidence from Peruvian gold mining that enforcement failures, rather than the absence of environmental laws, drive illegal extraction and associated damages. Their work implies that governments may deliberately relax enforcement during periods of high mining profitability, though direct quantitative evidence on policy adjustment remains scarce. Farzanegan and Markwardt (2018) examine Middle Eastern and North African countries and find that democratic institutions moderate local pollution but fail to constrain global pollutants like CO₂, suggesting that political regimes shape environmental outcomes in complex ways.

The financial sector also plays a mediating role. Ulucak, Danish, and Ozcan (2020) demonstrate that financial development interacts with natural resource rents to affect environmental sustainability in OECD countries, while Paramati, Ummalla, and Apergis (2016) show that stock market development influences clean energy consumption and emissions. These studies indicate that resource-driven environmental effects are conditional on broader economic structures. Wang et al. (2020) further complicate the picture by documenting that globalization and agriculture value-added moderate the resource-emissions nexus in G7 economies post-COP21, suggesting that international commitments may alter domestic policy responses to resource shocks.

Despite this rich evidence on environmental outcomes, a critical gap remains in understanding the specific policy mechanism linking resource dependence to environmental degradation. The literature overwhelmingly focuses on final pollution levels—CO₂ emissions, ecological footprints, or deforestation—rather than on the policy inputs that governments control. It remains unclear whether observed increases in pollution during price booms stem purely from expanded economic activity, as Oyekola et al. (2024) and Chen et al. (2024) suggest, or from deliberate relaxation of environmental regulations, as the qualitative work of Espin and Perz (2021) implies. This distinction matters for policy: if governments actively weaken regulations during booms, international climate agreements must address enforcement and policy stability, not just emissions targets.

This study addresses that gap by shifting the focus from environmental outcomes to policy inputs. I investigate whether government effectiveness conditions the relationship between mineral dependence and environmental tax revenues, testing whether institutional quality serves as a measurable channel in the resource-environment relationship. Following the institutional approach advocated by Barma, Kaiser, Le, and Viñuela (2012), I argue that strong government institutions shield environmental policy from mining sector pressure. By examining environmental tax revenue—a direct measure of fiscal commitment to pollution pricing—rather than emissions outcomes, this study provides quantitative evidence on whether institutional quality determines how mineral wealth translates into environmental policy.

## 1.2. Data Cleaning/Loading 

```python
# Section: Cleaning_xvariables
import pandas as pd
import numpy as np

# Replace 'xvariables.csv' with your file path if it's not in the same folder
df = pd.read_csv('xvariables.csv')

# Display the first few rows to verify
df.head()
# Assuming df is your loaded DataFrame

# Drop rows where Series Name is exactly "Tax revenue (% of GDP)"
df_filtered = df[df['Series Name'] != 'Tax revenue (% of GDP)'].copy()

# Optional: reset index if you want clean row numbers
df_filtered = df_filtered.reset_index(drop=True)

# 3. Convert year columns to numeric
year_cols = [col for col in df.columns if ' [YR' in col and col[-1].isdigit()]
for col in year_cols:
    df[col] = pd.to_numeric(df[col], errors='coerce')

# 4. Define target years: 2003 to 2020 inclusive
target_years = [f"{y} [YR{y}]" for y in range(2003, 2021)]

# Ensure all target years exist (rarely needed)
for y in target_years:
    if y not in df.columns:
        df[y] = np.nan

# 5. Step 1: Drop all "Tax revenue (% of GDP)" rows
df_no_tax = df[df['Series Name'] != 'Tax revenue (% of GDP)'].copy()

# 6. Step 2: Check completeness (no NaN allowed in any year 2003–2020)
#    → must be non-NaN in EVERY year for EVERY remaining series
is_complete = (
    df_no_tax.groupby(['Country Name', 'Series Name'])[target_years]
    .apply(lambda g: g.notna().all().all())   # True only if zero NaNs in the 18 years
    .reset_index(name='is_fully_complete')
)

# 7. Country is kept only if ALL its remaining series are fully complete
countries_completeness = (
    is_complete
    .groupby('Country Name')['is_fully_complete']
    .all()   # True only if every series for the country is True
    .reset_index(name='all_series_complete')
)

countries_to_keep = countries_completeness[
    countries_completeness['all_series_complete']
]['Country Name'].tolist()

# Clean and sort
countries_to_keep = sorted([str(c) for c in countries_to_keep if pd.notna(c)])

# 9. Final filtered DataFrame (only kept countries, tax revenue already dropped)
df_final = df_no_tax[df_no_tax['Country Name'].isin(countries_to_keep)].copy()

# 1. Check what variables exist (look for df_final, df_no_tax, etc.)
%who_ls DataFrame

# 1. Define the columns to keep
keep_columns = ['Country Name', 'Series Name']

# Add only the years we want: 2003 to 2020
for year in range(2003, 2021):
    col_name = f"{year} [YR{year}]"
    if col_name in df_final.columns:
        keep_columns.append(col_name)

# 2. Create the cleaned DataFrame
df_clean = df_final[keep_columns].copy()

# 3. Optional: sort by country and series for better readability
df_clean = df_clean.sort_values(['Country Name', 'Series Name']).reset_index(drop=True)

# Rename columns like "2003 [YR2003]" → "2003"
df_clean = df_clean.rename(
    columns=lambda col: col.split(' [YR')[0] if ' [YR' in col and col.endswith(']') else col
)

# Optional: make sure only the desired years remain (just in case)
years_to_keep = [str(y) for y in range(2003, 2021)]
columns_to_keep = ['Country Name', 'Series Name'] + years_to_keep

# Filter to only keep these columns (removes any leftover unwanted years)
df_clean = df_clean[[col for col in columns_to_keep if col in df.columns]]


# Assuming your cleaned DataFrame is called 'df_clean' (change the name if different)
df_clean.to_csv('cleaned_check_xvariables.csv', index=False)


# Section: Cleaning_y

# Load the CSV file (replace with your exact filename if different)
df = pd.read_csv('y_environmental_taxes.csv')


# Assuming 'df' is your current DataFrame after previous cleaning steps

# Keep only rows where:
# - Indicator is "Environmental Taxes"
# - Unit is "Percent of GDP"
df_gdp = df[
    (df['Indicator'] == 'Environmental Taxes') &
    (df['Unit'] == 'Percent of GDP')
].copy()

# Step 2: Now drop the columns you don't need
columns_to_drop = ['ObjectId'
    'ISO2', 'ISO3', 'Source',
    'CTS Code', 'CTS Full Descriptor',
    '1995', '1996', '1997', '1998', '1999',
    '2000', '2001', '2002']

df_gdp = df_gdp.drop(columns=columns_to_drop, errors='ignore')

# 3. Optional: sort by country and clean index
df_gdp = df_gdp.sort_values('Country').reset_index(drop=True)



# Assuming your current clean DataFrame is called df_gdp
# (the one that has 'Country' + year columns like '2003', '2004', ..., '2021')

# 1. Define the years we care about
years = [str(year) for year in range(2003, 2021)]   # 2003 to 2021 inclusive

# 2. Make sure all these year columns actually exist in the DataFrame
available_years = [y for y in years if y in df_gdp.columns]

# 3. Drop rows that have ANY NaN in any of these year columns
df_complete = df_gdp.dropna(subset=available_years).copy()

# 4. Reset index for clean look
df_complete = df_complete.reset_index(drop=True)


# Columns to keep: Country + all year columns (adjust if you need ISO2/ISO3 for merging later)
columns_to_keep = ['Country'] + [str(year) for year in range(2003, 2021)]

# Or if you want to keep ISO2/ISO3:
# columns_to_keep = ['Country', 'ISO2', 'ISO3'] + [str(year) for year in range(2003, 2022)]

df_complete = df_complete[columns_to_keep].copy()

# Preview to confirm

df_complete.to_csv('cleaned_check_yvariable.csv', index=False)
```

```python
# Merging 
import nbconvert
x_df = pd.read_csv('cleaned_xvariables.csv')
y_df = pd.read_csv('cleaned_check_yvariable.csv')

# Define the mapping (from y-variable names to x-variable names)
name_mapping = {
    'China, P.R.: Mainland': 'China',
    'Congo, Dem. Rep. of the': 'Congo, Dem. Rep.',
    'Congo, Rep. of': 'Congo, Rep.',
    'Croatia, Rep. of': 'Croatia',
    'Czech Rep.': 'Czechia',
    'Egypt, Arab Rep. of': 'Egypt, Arab Rep.',
    'Estonia, Rep. of': 'Estonia',
    'Kyrgyz Rep.': 'Kyrgyz Republic',
    'Netherlands, The': 'Netherlands',
    'Poland, Rep. of': 'Poland',
    'Slovak Rep.': 'Slovak Republic',
    'Slovenia, Rep. of': 'Slovenia',
    'Türkiye, Rep. of': 'Turkiye',
    'Vietnam': 'Viet Nam',
}

# Load the data
y_df = pd.read_csv('cleaned_check_yvariable.csv')
x_df = pd.read_csv('cleaned_xvariables.csv')

# Apply the mapping to standardize country names in y_df
y_df['Country'] = y_df['Country'].replace(name_mapping)

# Melt y to long format
y_long = pd.melt(
    y_df, 
    id_vars=['Country'], 
    var_name='year', 
    value_name='environmental_taxes'
)
y_long['year'] = y_long['year'].astype(int)

# Melt x to long format
x_long = pd.melt(
    x_df, 
    id_vars=['Country Name', 'Series Name'], 
    var_name='year', 
    value_name='value'
)
x_long['year'] = x_long['year'].astype(int)

# Pivot x so each series becomes a column
x_pivoted = x_long.pivot(
    index=['Country Name', 'year'], 
    columns='Series Name', 
    values='value'
).reset_index()

# Rename for consistency
x_pivoted.rename(columns={'Country Name': 'Country'}, inplace=True)

# Merge on standardized 'Country' and 'year'
merged_df = pd.merge(
    y_long, 
    x_pivoted, 
    on=['Country', 'year'], 
    how='inner'  # Keeps only matching country-years
)

# Save the merged dataset
merged_df.to_csv('merged_dataset_standardized.csv', index=False)


# Save the full merged dataset as CSV
merged_df.to_csv('merged_environmental_taxes_and_controls.csv', index=False)

# Load your current merged file (X + Y + controls)
df = pd.read_csv('merged_environmental_taxes_and_controls.csv')

df

```

Figure 3: Stronger Government Institutions Are Associated with Higher Environmental Taxes

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Load your data
df = pd.read_csv('final_panel_with_boom.csv')

# ==================== NEW CLEAN PLOT: Y vs Government Effectiveness ====================
# This is your requested "governance" plot (distinct control variable)
# Simple regplot – exactly like your good GDP-per-capita plot (no skew, no zero-cluster noise)

plt.figure(figsize=(10, 6))
sns.regplot(data=df, x='Government Effectiveness: Estimate', y='environmental_taxes',
            scatter_kws={'alpha': 0.45, 's': 18}, line_kws={'color': 'green'}, ci=95)

plt.title('Environmental Tax Revenues vs Government Effectiveness\n(Control Relationship)', fontsize=14, pad=15)
plt.xlabel('Government Effectiveness: Estimate (World Bank WGI)', fontsize=12)
plt.ylabel('Environmental Taxes (% of GDP)', fontsize=12)
plt.grid(True, alpha=0.3)
plt.tight_layout()

# Save it (ready for your project)
plt.savefig('plot_governance_Y_vs_government_effectiveness.png', dpi=300)
plt.show()
```

Government effectiveness, measured by the World Bank’s Worldwide Governance Indicators (WGI) estimate, serves as the central institutional mechanism in this study. It is hypothesized to moderate the relationship between mineral dependence and environmental policy stringency. As shown in the figure above, there is a strong positive association between government effectiveness and environmental tax revenues as a share of GDP. The upward-sloping regression line and narrow confidence band indicate that countries with higher-quality public institutions and more competent bureaucracies tend to collect significantly higher environmental taxes.
This descriptive pattern is consistent with the theoretical argument that effective governments possess greater administrative capacity and political credibility to design, implement, and enforce environmental taxation, even in the face of potential opposition from resource interests. The positive relationship also provides preliminary support for the conditioning hypothesis: stronger institutions may enable mineral-dependent economies to escape the tendency toward weaker environmental regulations. This motivates the main empirical analysis, which formally tests whether government effectiveness attenuates the expected negative effect of mineral rents on environmental tax collection.

Figure 4: Average Environmental Taxes over Time – Line Plot by Mineral Dependence Group

Purpose: Provide insight into dynamic behavior and potential differential trends around the commodity boom period (~2003–2014), using a grouping that proxies exposure to mechanism Z.

```python
# Create the group indicator
median_rent = df['Mineral rents (% of GDP)'].median()
df['High mineral dependence'] = df['Mineral rents (% of GDP)'] > median_rent

# Group by the REAL year column name — change 'year' to the actual name!
yearly_means = df.groupby(['year', 'High mineral dependence'])['environmental_taxes'].mean().unstack()

# Plot
import matplotlib.pyplot as plt

plt.figure(figsize=(11, 6))
yearly_means.plot(marker='o', linewidth=2)
plt.title('Average Environmental Taxes over Time\nby Mineral Dependence Group', fontsize=13)
plt.xlabel('Year')
plt.ylabel('Environmental taxes (% of GDP)')
plt.legend(['Low mineral rents', 'High mineral rents'], title='Group')
plt.axvspan(2003, 2014, alpha=0.08, color='gray', label='Commodity boom period (approx.)')
plt.grid(True, alpha=0.3)
plt.tight_layout()
plt.show()
```

The figure reveals two consistent patterns. First, countries with low mineral dependence (blue line) consistently exhibit higher average environmental tax revenues throughout the entire sample period compared to high mineral-dependent countries (orange line). Second, the gap between the two groups remains relatively stable over time, with only modest convergence during the peak of the commodity boom (approximately 2003–2014, shaded region).
These trends provide descriptive evidence supporting the central hypothesis of the study: greater economic dependence on mineral rents is associated with persistently weaker environmental policy outcomes, as proxied by lower environmental tax collection. The fact that the divergence does not appear to be driven solely by the commodity boom period suggests that mineral dependence may reflect a structural, rather than cyclical, constraint on environmental regulation. This motivates the subsequent analysis, which investigates whether strong government institutions can help mineral-dependent countries overcome this tendency and achieve environmental tax revenues closer to those observed in less resource-reliant economies.
